```python
import torch
from torch import nn
import numpy as np
import pandas as pd
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
x = torch.tensor(5.0, requires_grad=True)
f = x ** 2
f
```

```python
f.backward()
x.grad
```

```python
# After computing the gradient, you usually perform gradient descent, by subtracting a fraction of the gradients from the model variables (usually done in neural networks).
# To do a gradient descent step, you must temporarily disable gradient tracking, since you don't want to track the gradient descent step itself in the computation graph.
# This can be done by placing the gradient descent step inside a torch.no_grad() context:
learning_rate = 0.1
with torch.no_grad():
    x -= learning_rate * x.grad
x
```

```python
# Another way of doing this is using the variable's detach() method, this will create a new tensor detached from the computation graph, with requires_grad = False.
# This new tensor shares the same data in memory, so the original tensor will be updated.
x_detached = x.detach()
x_detached -= learning_rate * x.grad
x
```

```python
# Before repeating the whole process on training, you need to zero out the gradients of every model parameter:
x.grad.zero_()
x.grad
```

```python
# Putting everything together:
learning_rate = 0.1
x = torch.tensor(5.0, requires_grad=True)
for iteration in range(100):
    f = x ** 2
    f.backward()
    with torch.no_grad():
        x -= learning_rate * x.grad # gradient descent step
        x.grad.zero_()              # Zero out gradient before next iteration
x
```
